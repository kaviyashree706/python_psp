<h2 style="color:#2E7D32;">Question</h2>

Write a Python program to connect to an authorized network service and display the service information provided by it.

<h2 style="color:#E65100;">Aim</h2>

To connect to an authorized network service and retrieve the banner or service information provided by the server.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Import the <b>socket</b> module.
3. Enter the authorized server address and port number.
4. Create a TCP socket.
5. Connect to the server.
6. Receive the service information from the server.
7. Decode and display the received information.
8. Close the connection.
9. Stop the program.

```python
import socket

host = input("Enter authorized server address: ")
port = int(input("Enter port number: "))

try:
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.settimeout(5)

    s.connect((host, port))

    banner = s.recv(1024)

    print("\nService Information:")
    print(banner.decode(errors="ignore"))

    s.close()

except Exception as e:
    print("Unable to connect:", e)
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>import socket</b> imports the socket module for network communication.

• <b>input()</b> gets the authorized server address and port number.

• <b>socket()</b> creates a TCP socket.

• <b>settimeout()</b> prevents the program from waiting too long for a response.

• <b>connect()</b> connects to the authorized server.

• <b>recv()</b> receives service information from the server.

• <b>decode()</b> converts the received data into readable text.

• <b>try</b> and <b>except</b> handle connection errors safely.
