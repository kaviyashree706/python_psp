<h1 style="color:#1565C0;">04_HTTP GET Request</h1>

<h2 style="color:#2E7D32;">Question</h2>

Write a Python program to get the contents of a given webpage using an HTTP GET request and display the response.

<h2 style="color:#E65100;">Aim</h2>

To retrieve the contents of a webpage using an HTTP GET request and display the response.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Import the <b>requests</b> module.
3. Get the webpage URL from the user.
4. Send an HTTP GET request to the webpage.
5. Receive the response from the server.
6. Check the response status.
7. Display the webpage contents.
8. Stop the program.

```python
import requests

url = input("Enter webpage URL: ")

response = requests.get(url)

print("\nStatus Code:", response.status_code)
print("\nWebpage Content:")
print(response.text[:1000])
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>import requests</b> imports the requests module for sending HTTP requests.

• <b>input()</b> gets the webpage URL from the user.

• <b>requests.get()</b> sends an HTTP GET request to the webpage.

• <b>response.status_code</b> shows the response status from the server.

• <b>response.text</b> contains the webpage content.

• <b>[:1000]</b> displays only the first 1000 characters of the response.
