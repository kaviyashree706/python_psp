<h1 style="color:#1565C0;">14_Buffer Overflow Simulation</h1>

<h2 style="color:#2E7D32;">Question</h2>

Create a safe program that accepts user input, checks its length against a fixed buffer size, and shows what happens when the input is too large.

<h2 style="color:#E65100;">Aim</h2>

To understand the basic idea of a buffer overflow by checking whether the user input exceeds a fixed buffer size.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Set a fixed buffer size.
3. Get input from the user.
4. Find the length of the input.
5. Compare the input length with the buffer size.
6. If the input is within the limit, accept it.
7. If the input is larger than the limit, reject it and display a warning.
8. Stop the program.

```python
BUFFER_SIZE = 10

data = input("Enter data: ")

print("\nBuffer Size:", BUFFER_SIZE)
print("Input Length:", len(data))

if len(data) <= BUFFER_SIZE:
    print("Input accepted.")
    print("Data:", data)
else:
    print("Input too large!")
    print("Buffer Overflow prevented.")
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>BUFFER_SIZE = 10</b> sets the maximum allowed input size.

• <b>input()</b> gets data from the user.

• <b>len(data)</b> finds the number of characters entered.

• The <b>if</b> condition checks whether the input is within the buffer limit.

• If the input is within the limit, the program accepts it.

• If the input is larger than the limit, the program rejects it and displays a warning.

• This is a safe simulation because the program does not write beyond a real memory buffer.
