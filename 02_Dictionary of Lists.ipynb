<h1 style="color:#1565C0;">02_Dictionary of Lists</h1>

<h2 style="color:#2E7D32;">Question</h2>

Create a dictionary to store each student name along with their marks in three subjects. Display the student details and marks.

<h2 style="color:#E65100;">Aim</h2>

To create a dictionary containing student names and their marks in three subjects and display the stored details.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Create an empty dictionary.
3. Get the student name and marks in three subjects.
4. Store the marks as a list for each student.
5. Add the student details to the dictionary.
6. Display the student names and their marks.
7. Stop the program.

```python
students = {}

n = int(input("Enter number of students: "))

for i in range(n):
    name = input("\nEnter student name: ")
    
    mark1 = int(input("Enter mark for Subject 1: "))
    mark2 = int(input("Enter mark for Subject 2: "))
    mark3 = int(input("Enter mark for Subject 3: "))
    
    students[name] = [mark1, mark2, mark3]

print("\nStudent Details:")

for name, marks in students.items():
    print("Student Name:", name)
    print("Marks:", marks)
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>students = {}</b> creates an empty dictionary.

• <b>input()</b> gets the student name and marks.

• <b>int()</b> converts the entered marks into numbers.

• The three marks are stored together as a <b>list</b>.

• <b>students[name] = [mark1, mark2, mark3]</b> stores the student's name and marks in the dictionary.

• <b>for</b> loop is used to display all student details.

• <b>items()</b> gets the student name and marks from the dictionary.
