<h1 style="color:#1565C0;">20_Buffer Overflow Protection</h1>

<h2 style="color:#2E7D32;">Question</h2>

Write a program that accepts user input only when its length is within the allowed limit and rejects oversized input safely.

<h2 style="color:#E65100;">Aim</h2>

To protect a program from oversized input by checking the input length before accepting it.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Set a maximum allowed input length.
3. Get input from the user.
4. Find the length of the input.
5. Compare the input length with the allowed limit.
6. If the input is within the limit, accept and display it.
7. If the input exceeds the limit, reject it safely.
8. Display the appropriate message.
9. Stop the program.

```python
MAX_LENGTH = 10

user_input = input("Enter your data: ")

if len(user_input) <= MAX_LENGTH:
    print("\nInput accepted.")
    print("Data:", user_input)
else:
    print("\nInput rejected.")
    print("Error: Input exceeds the allowed limit.")
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>MAX_LENGTH = 10</b> sets the maximum number of characters allowed.

• <b>input()</b> gets data from the user.

• <b>len()</b> calculates the length of the entered data.

• The <b>if</b> condition checks whether the input is within the allowed limit.

• If the input is within the limit, it is accepted.

• If the input is too large, it is rejected safely.

• This length checking helps prevent oversized input from being processed.

Data: Hello
