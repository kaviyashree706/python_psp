<h1 style="color:#1565C0;">12_Ping Sweep</h1>

<h2 style="color:#2E7D32;">Question</h2>

Write a Python program to check a given range of authorized IP addresses and display which hosts are active.

<h2 style="color:#E65100;">Aim</h2>

To check a range of authorized IP addresses and identify which hosts are reachable.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Enter the authorized network range.
3. Generate the IP addresses in the given range.
4. Send a ping request to each IP address.
5. Check whether the host responds.
6. Display the active hosts.
7. Stop the program.

```python
import subprocess
import ipaddress

network = input("Enter authorized network (example: 192.168.1.0/29): ")

try:
    net = ipaddress.ip_network(network, strict=False)

    print("\nActive Hosts:")

    for ip in net.hosts():
        result = subprocess.run(
            ["ping", "-n", "1", "-w", "1000", str(ip)],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL
        )

        if result.returncode == 0:
            print(ip, "is Active")

except ValueError:
    print("Invalid network address.")
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>subprocess</b> is used to run the system ping command.

• <b>ipaddress</b> is used to generate IP addresses from the given network range.

• <b>input()</b> gets the authorized network range from the user.

• <b>net.hosts()</b> generates the usable host IP addresses.

• <b>ping</b> checks whether each host is reachable.

• <b>returncode == 0</b> means the host responded successfully.

• <b>print()</b> displays the active hosts.
