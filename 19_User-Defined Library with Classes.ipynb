<h1 style="color:#1565C0;">19_User-Defined Library with Classes</h1>

<h2 style="color:#2E7D32;">Question</h2>

Create a user-defined library containing a <b>Student</b> class. Import the library into another program and display student information.

<h2 style="color:#E65100;">Aim</h2>

To create a user-defined library with a Student class and import it into another Python program.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Create a Python file named <b>student.py</b>.
3. Define a <b>Student</b> class inside the file.
4. Create a constructor to store student name and marks.
5. Create a method to display student information.
6. Create another Python program.
7. Import the <b>Student</b> class from the library.
8. Create a Student object.
9. Display the student information.
10. Stop the program.

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Student Name:", self.name)
        print("Marks:", self.marks)
```

```python
from student import Student

name = input("Enter student name: ")
marks = int(input("Enter student marks: "))

student = Student(name, marks)

print("\nStudent Information:")
student.display()
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>student.py</b> is the user-defined library file.

• <b>Student</b> is a class created inside the library.

• <b>__init__()</b> stores the student's name and marks.

• <b>display()</b> displays the student information.

• <b>from student import Student</b> imports the Student class into the main program.

• <b>Student(name, marks)</b> creates an object of the Student class.

• <b>student.display()</b> calls the display method.
