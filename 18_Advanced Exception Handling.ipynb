<h2 style="color:#2E7D32;">Question</h2>

Write a Python program that performs division and handles different errors such as invalid input and division by zero using appropriate exception blocks.

<h2 style="color:#E65100;">Aim</h2>

To perform division and handle different types of errors using appropriate exception handling blocks.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Get two numbers from the user.
3. Convert the input values into numbers.
4. Perform division of the first number by the second number.
5. Handle invalid input using <b>ValueError</b>.
6. Handle division by zero using <b>ZeroDivisionError</b>.
7. Use <b>finally</b> to display a completion message.
8. Stop the program.

```python
try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    result = num1 / num2

    print("Result:", result)

except ValueError:
    print("Invalid input! Please enter numbers only.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

except Exception as e:
    print("An unexpected error occurred:", e)

finally:
    print("Division operation completed.")
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>try</b> contains the code that may produce an error.

• <b>float()</b> converts the entered values into numbers.

• <b>ValueError</b> handles invalid input such as letters instead of numbers.

• <b>ZeroDivisionError</b> handles the situation when the second number is zero.

• <b>Exception</b> handles any other unexpected error.

• <b>finally</b> executes whether an error occurs or not.

• <b>print()</b> displays the result or the appropriate error message.
