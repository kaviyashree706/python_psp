<h2 style="color:#2E7D32;">Question</h2>

Create a socket server that waits for a client, receives a message, and sends a reply back to the client.

<h2 style="color:#E65100;">Aim</h2>

To create a simple socket server that receives a message from a client and sends a response back.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Import the <b>socket</b> module.
3. Create a TCP socket for the server.
4. Bind the server to localhost and a port number.
5. Listen for a client connection.
6. Accept the client connection.
7. Receive a message from the client.
8. Display the received message.
9. Send a reply to the client.
10. Close the connection.
11. Stop the program.

```python
import socket

server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

server.bind(("localhost", 9999))
server.listen(1)

print("Server is waiting for a client...")

client, address = server.accept()

message = client.recv(1024).decode()

print("Message received:", message)

reply = "Message received successfully!"
client.send(reply.encode())

client.close()
server.close()
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>import socket</b> imports the socket module for network communication.

• <b>socket()</b> creates a TCP socket.

• <b>bind()</b> assigns the server to localhost and port 9999.

• <b>listen()</b> makes the server wait for a client connection.

• <b>accept()</b> accepts a connection from the client.

• <b>recv()</b> receives the message sent by the client.

• <b>send()</b> sends a reply back to the client.

• <b>close()</b> closes the client and server connections.
