<h1 style="color:#1565C0;">17_HTTP Client with Sockets</h1>

<h2 style="color:#2E7D32;">Question</h2>

Create an HTTP client using Python sockets that connects to a test web server, sends a request, and displays the response.

<h2 style="color:#E65100;">Aim</h2>

To create an HTTP client using Python sockets and retrieve a response from a test web server.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Import the <b>socket</b> module.
3. Enter the test web server address.
4. Create a TCP socket.
5. Connect the socket to the web server.
6. Create an HTTP GET request.
7. Send the request to the server.
8. Receive the response from the server.
9. Display the response.
10. Close the socket.
11. Stop the program.

```python
import socket

host = input("Enter test web server: ")

client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
client.connect((host, 80))

request = "GET / HTTP/1.1\r\nHost: " + host + "\r\nConnection: close\r\n\r\n"

client.send(request.encode())

response = b""

while True:
    data = client.recv(4096)

    if not data:
        break

    response += data

client.close()

print("\nHTTP Response:")
print(response.decode(errors="ignore")[:2000])
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>import socket</b> imports the socket module for network communication.

• <b>input()</b> gets the test web server address from the user.

• <b>socket()</b> creates a TCP socket.

• <b>connect()</b> connects the client to the web server on port 80.

• <b>GET /</b> creates an HTTP GET request for the main webpage.

• <b>send()</b> sends the HTTP request to the server.

• <b>recv()</b> receives the response from the server.

• <b>decode()</b> converts the received data into readable text.

• <b>close()</b> closes the socket connection.
