<h1 style="color:#1565C0;">09_Using Global Variables in Functions</h1>

<h2 style="color:#2E7D32;">Question</h2>

Create a global variable called <b>count</b> and write a function to display and update its value.

<h2 style="color:#E65100;">Aim</h2>

To create and use a global variable inside a function and update its value.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Create a global variable called <b>count</b> and assign it a value.
3. Define a function named <b>update_count()</b>.
4. Use the <b>global</b> keyword to access the global variable.
5. Display the current value of <b>count</b>.
6. Increase the value of <b>count</b> by 1.
7. Display the updated value.
8. Call the function.
9. Stop the program.

```python
count = 0

def update_count():
    global count

    print("Current count:", count)
    
    count = count + 1
    
    print("Updated count:", count)

update_count()
update_count()
update_count()
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>count = 0</b> creates a global variable named <b>count</b>.

• <b>def update_count()</b> defines a function to display and update the count.

• <b>global count</b> allows the function to change the global variable.

• <b>count = count + 1</b> increases the count value by 1.

• The function is called three times, so the value increases each time.

<h2 style="color:#C62828;">Output</h2>

<b>Example Output:</b>

Current count: 0
Updated count: 1

Current count: 1
Updated count: 2

Current count: 2
Updated count: 3
