<h1 style="color:#1565C0;">05_Retrieving Host Details</h1>

<h2 style="color:#2E7D32;">Question</h2>

Write a Python program to find the computer's hostname and IP address and display them.

<h2 style="color:#E65100;">Aim</h2>

To retrieve and display the hostname and IP address of the computer using Python.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Import the <b>socket</b> module.
3. Get the hostname of the computer.
4. Get the IP address using the hostname.
5. Display the hostname and IP address.
6. Stop the program.

```python
import socket

hostname = socket.gethostname()
ip_address = socket.gethostbyname(hostname)

print("Host Name:", hostname)
print("IP Address:", ip_address)
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>import socket</b> imports the socket module for network-related operations.

• <b>socket.gethostname()</b> gets the name of the computer.

• <b>socket.gethostbyname()</b> gets the IP address of the computer using its hostname.

• <b>print()</b> displays the hostname and IP address.

<h2 style="color:#C62828;">Output</h2>

<b>Example Output:</b>

Host Name: DESKTOP-ABC123
IP Address: 192.168.1.10
