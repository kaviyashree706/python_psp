<h1 style="color:#1565C0;">01_File Handling and Exception Handling</h1>

<h2 style="color:#2E7D32;">Question</h2>

Create a file named <b>student.txt</b>, write student details into it, read the details, and display a proper message if an error occurs.

<h2 style="color:#E65100;">Aim</h2>

To create a file, write student details into the file, read the details, and handle errors using exception handling.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Get the student name and marks from the user.
3. Create and open <b>student.txt</b> in write mode.
4. Write the student details into the file.
5. Close the file.
6. Open the file in read mode.
7. Read and display the student details.
8. Handle file errors using <b>try</b> and <b>except</b>.
9. Stop the program.

```python
try:
    name = input("Enter student name: ")
    marks = input("Enter student marks: ")

    file = open("student.txt", "w")
    file.write("Student Name: " + name + "\n")
    file.write("Marks: " + marks + "\n")
    file.close()

    file = open("student.txt", "r")
    data = file.read()
    file.close()

    print("\nStudent Details:")
    print(data)

except FileNotFoundError:
    print("File not found.")

except Exception as e:
    print("An error occurred:", e)
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>input()</b> gets the student's name and marks.

• <b>open("student.txt", "w")</b> creates the file and opens it in write mode.

• <b>write()</b> stores the student details in the file.

• <b>open("student.txt", "r")</b> opens the file in read mode.

• <b>read()</b> reads the contents of the file.

• <b>print()</b> displays the student details.

• <b>try</b> is used to perform the file operations safely.

• <b>except</b> is used to handle errors.
