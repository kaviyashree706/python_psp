<h1 style="color:#1565C0;">03_UDP Server and Client</h1>

<h2 style="color:#2E7D32;">Question</h2>

Create a UDP client and server where the client sends a student's name and the server receives and displays it.

<h2 style="color:#E65100;">Aim</h2>

To create a UDP client and server and send a student's name from the client to the server.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Create a UDP socket for the server.
3. Bind the server to an IP address and port number.
4. Create a UDP socket for the client.
5. Get the student's name from the client.
6. Send the name from the client to the server.
7. Receive the name at the server.
8. Display the received student name.
9. Stop the program.

```python
import socket

# Create UDP server
server = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
server.bind(("localhost", 9999))

print("UDP Server is waiting...")

# Create UDP client
client = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

name = input("Enter student name: ")

# Send student name to server
client.sendto(name.encode(), ("localhost", 9999))

# Receive data at server
data, address = server.recvfrom(1024)

print("\nStudent Name received by server:", data.decode())

client.close()
server.close()
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>socket.socket()</b> creates a socket for communication.

• <b>SOCK_DGRAM</b> is used to create a UDP socket.

• <b>bind()</b> connects the server to the localhost and port number.

• <b>input()</b> gets the student's name from the user.

• <b>sendto()</b> sends the student's name from the client to the server.

• <b>encode()</b> converts the name into bytes for sending.

• <b>recvfrom()</b> receives the data at the server.

• <b>decode()</b> converts the received bytes back into text.
