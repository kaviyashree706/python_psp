<h1 style="color:#1565C0;">08_Exception Handling in File Operations</h1>

<h2 style="color:#2E7D32;">Question</h2>

Write a Python program to open and read a file. If the file does not exist, display <b>"File not found"</b> instead of showing an error.

<h2 style="color:#E65100;">Aim</h2>

To read the contents of a file and handle the file-not-found error using exception handling.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Get the file name from the user.
3. Open the file in read mode.
4. Read the contents of the file.
5. Display the file contents.
6. If the file does not exist, handle the error using <b>FileNotFoundError</b>.
7. Display <b>"File not found"</b>.
8. Stop the program.

```python
try:
    filename = input("Enter file name: ")

    file = open(filename, "r")
    data = file.read()

    print("\nFile Contents:")
    print(data)

    file.close()

except FileNotFoundError:
    print("File not found")
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>input()</b> gets the file name from the user.

• <b>open(filename, "r")</b> opens the file in read mode.

• <b>read()</b> reads the contents of the file.

• <b>print()</b> displays the file contents.

• <b>try</b> is used to execute the file operation safely.

• <b>except FileNotFoundError</b> handles the error when the file does not exist.

• If the file is not found, the program displays <b>"File not found"</b> instead of stopping with an error.

<h2 style="color:#C62828;">Output</h2>

<b>Example 1 — File Exists:</b>

Enter file name: student.txt

File Contents:

<b>Example 2 — File Does Not Exist:</b>
