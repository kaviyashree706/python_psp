<h2 style="color:#2E7D32;">Question</h2>

Using Scapy, capture packets from your own test network and display basic information such as source, destination, and protocol.

<h2 style="color:#E65100;">Aim</h2>

To capture packets from an authorized test network using Scapy and display basic packet information.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Import the required Scapy functions.
3. Capture a small number of packets from the authorized test network.
4. Check whether each packet contains IP information.
5. Get the source IP address.
6. Get the destination IP address.
7. Identify the protocol.
8. Display the packet information.
9. Stop the program.

```python
from scapy.all import sniff, IP

def display_packet(packet):
    if IP in packet:
        source = packet[IP].src
        destination = packet[IP].dst
        protocol = packet[IP].proto

        print("Source:", source)
        print("Destination:", destination)
        print("Protocol:", protocol)
        print("-" * 40)

print("Capturing packets from the authorized test network...")

sniff(prn=display_packet, count=5)
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>from scapy.all import sniff, IP</b> imports the required Scapy functions.

• <b>sniff()</b> captures network packets.

• <b>count=5</b> limits the capture to 5 packets.

• <b>IP in packet</b> checks whether the packet contains IP information.

• <b>packet[IP].src</b> gets the source IP address.

• <b>packet[IP].dst</b> gets the destination IP address.

• <b>packet[IP].proto</b> gets the protocol number.

• <b>display_packet()</b> displays the basic information of each captured packet.
