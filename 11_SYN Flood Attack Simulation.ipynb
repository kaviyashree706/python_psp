<h1 style="color:#1565C0;">11_SYN Flood Attack Simulation</h1>

<h2 style="color:#2E7D32;">Question</h2>

Demonstrate the basic idea of a SYN flood using a safe lab simulation and explain how repeated connection requests can affect a server.

<h2 style="color:#E65100;">Aim</h2>

To understand the basic working of a SYN flood through a safe simulation and observe how repeated connection requests can use server resources.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Create a safe local server simulation.
3. Generate a small number of simulated connection requests.
4. Store the requests as pending connections.
5. Display the number of pending requests.
6. Explain that a large number of pending requests can consume server resources.
7. Do not send traffic to real or unauthorized systems.
8. Stop the simulation.

```python
# Safe SYN Flood Simulation
# This program only simulates connection requests.
# It does not send real network traffic.

pending_requests = []

number_of_requests = int(input("Enter number of simulated SYN requests: "))

for i in range(1, number_of_requests + 1):
    request = "SYN Request " + str(i)
    pending_requests.append(request)

print("\nSYN Flood Simulation")
print("--------------------")

for request in pending_requests:
    print(request)

print("\nTotal Pending Requests:", len(pending_requests))

if len(pending_requests) > 10:
    print("Server resource usage may increase.")
else:
    print("Small number of requests - server can handle the simulation.")
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>pending_requests</b> stores the simulated connection requests.

• <b>input()</b> gets the number of requests to simulate.

• <b>for loop</b> creates the requested number of simulated SYN requests.

• <b>append()</b> adds each simulated request to the list.

• <b>len()</b> counts the total number of pending requests.

• If many requests are pending, the program displays that server resource usage may increase.

• This is only a <b>local simulation</b>; it does not create or send real SYN flood traffic.
