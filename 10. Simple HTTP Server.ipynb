<h2 style="color:#2E7D32;">Question</h2>

Create a simple Python web server that displays <b>"Welcome to My Server"</b> when opened in a browser.

<h2 style="color:#E65100;">Aim</h2>

To create a simple HTTP web server using Python and display a welcome message in a web browser.

<h2 style="color:#6A1B9A;">Algorithm</h2>

1. Start the program.
2. Import the required HTTP server modules.
3. Create a simple request handler.
4. Start the HTTP server on localhost using a port number.
5. Display a message when the server starts.
6. Open the server address in a web browser.
7. Display <b>"Welcome to My Server"</b>.
8. Stop the server when required.

```python
from http.server import BaseHTTPRequestHandler, HTTPServer

class MyServer(BaseHTTPRequestHandler):

    def do_GET(self):
        self.send_response(200)
        self.send_header("Content-type", "text/html")
        self.end_headers()

        message = "<h1>Welcome to My Server</h1>"
        self.wfile.write(message.encode())

server = HTTPServer(("localhost", 8000), MyServer)

print("Server started at http://localhost:8000")

server.serve_forever()
```

<h2 style="color:#00838F;">Code Explanation</h2>

• <b>BaseHTTPRequestHandler</b> is used to handle HTTP requests.

• <b>HTTPServer</b> is used to create the web server.

• <b>do_GET()</b> handles a GET request from the browser.

• <b>send_response(200)</b> sends a successful response to the browser.

• <b>send_header()</b> specifies that the response contains HTML.

• <b>wfile.write()</b> sends the welcome message to the browser.

• <b>localhost</b> means the server is running on the same computer.

• <b>port 8000</b> is used to access the server.
